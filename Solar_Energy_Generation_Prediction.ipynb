## 1. Import Required Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

print("Libraries imported successfully!")
```

## 2. Create Solar Energy Dataset

```python
np.random.seed(42)

n = 500

data = pd.DataFrame({
    "Temperature": np.random.uniform(15, 45, n),
    "Humidity": np.random.uniform(20, 90, n),
    "Solar_Radiation": np.random.uniform(100, 1000, n),
    "Wind_Speed": np.random.uniform(0, 15, n),
    "Cloud_Cover": np.random.uniform(0, 100, n)
})

# Simulated solar energy generation
data["Energy_Generation"] = (
    0.35 * data["Solar_Radiation"]
    + 2 * data["Temperature"]
    - 0.5 * data["Humidity"]
    + 1.5 * data["Wind_Speed"]
    - 1.2 * data["Cloud_Cover"]
    + np.random.normal(0, 30, n)
)

data.head()
```

## 3. Explore the Dataset

```python
print("Dataset Shape:", data.shape)

print("\nColumn Names:")
print(list(data.columns))

print("\nMissing Values:")
print(data.isnull().sum())

print("\nDataset Information:")
data.info()
```

## 4. Statistical Analysis

```python
data.describe()
```

## 5. Solar Radiation vs Energy Generation

```python
plt.figure(figsize=(8, 5))
plt.scatter(data["Solar_Radiation"], data["Energy_Generation"])
plt.xlabel("Solar Radiation")
plt.ylabel("Energy Generation")
plt.title("Solar Radiation vs Energy Generation")
plt.show()
```

## 6. Temperature vs Energy Generation

```python
plt.figure(figsize=(8, 5))
plt.scatter(data["Temperature"], data["Energy_Generation"])
plt.xlabel("Temperature")
plt.ylabel("Energy Generation")
plt.title("Temperature vs Energy Generation")
plt.show()
```

## 7. Humidity vs Energy Generation

```python
plt.figure(figsize=(8, 5))
plt.scatter(data["Humidity"], data["Energy_Generation"])
plt.xlabel("Humidity")
plt.ylabel("Energy Generation")
plt.title("Humidity vs Energy Generation")
plt.show()
```

## 8. Prepare Data for Machine Learning

```python
X = data[
    [
        "Temperature",
        "Humidity",
        "Solar_Radiation",
        "Wind_Speed",
        "Cloud_Cover"
    ]
]

y = data["Energy_Generation"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42
)

print("Training Data:", X_train.shape)
print("Testing Data:", X_test.shape)
```

## 9. Linear Regression Model

```python
linear_model = LinearRegression()
linear_model.fit(X_train, y_train)

linear_prediction = linear_model.predict(X_test)

linear_mae = mean_absolute_error(y_test, linear_prediction)
linear_rmse = np.sqrt(mean_squared_error(y_test, linear_prediction))
linear_r2 = r2_score(y_test, linear_prediction)

print("Linear Regression Results")
print("-------------------------")
print("MAE :", round(linear_mae, 2))
print("RMSE:", round(linear_rmse, 2))
print("R²  :", round(linear_r2, 4))
```

## 10. Random Forest Regression Model

```python
rf_model = RandomForestRegressor(
    n_estimators=100,
    random_state=42
)

rf_model.fit(X_train, y_train)

rf_prediction = rf_model.predict(X_test)

rf_mae = mean_absolute_error(y_test, rf_prediction)
rf_rmse = np.sqrt(mean_squared_error(y_test, rf_prediction))
rf_r2 = r2_score(y_test, rf_prediction)

print("Random Forest Regression Results")
print("--------------------------------")
print("MAE :", round(rf_mae, 2))
print("RMSE:", round(rf_rmse, 2))
print("R²  :", round(rf_r2, 4))
```

## 11. Compare Model Performance

```python
results = pd.DataFrame({
    "Model": ["Linear Regression", "Random Forest"],
    "MAE": [linear_mae, rf_mae],
    "RMSE": [linear_rmse, rf_rmse],
    "R2 Score": [linear_r2, rf_r2]
})

results.round(3)
```

## 12. Actual vs Predicted Solar Energy

```python
plt.figure(figsize=(8, 5))
plt.scatter(y_test, rf_prediction)

plt.xlabel("Actual Energy Generation")
plt.ylabel("Predicted Energy Generation")
plt.title("Actual vs Predicted Solar Energy")

plt.show()
```

## 13. Feature Importance

```python
importance = pd.Series(
    rf_model.feature_importances_,
    index=X.columns
).sort_values(ascending=False)

plt.figure(figsize=(8, 5))
importance.plot(kind="bar")

plt.xlabel("Features")
plt.ylabel("Importance")
plt.title("Feature Importance in Solar Energy Generation")
plt.xticks(rotation=45)

plt.show()

print("Feature Importance:")
print(importance.round(4))
```

## 14. Predict Solar Energy for New Weather Conditions

```python
# Enter weather conditions here

temperature = 30
humidity = 50
solar_radiation = 800
wind_speed = 5
cloud_cover = 20

new_data = pd.DataFrame({
    "Temperature": [temperature],
    "Humidity": [humidity],
    "Solar_Radiation": [solar_radiation],
    "Wind_Speed": [wind_speed],
    "Cloud_Cover": [cloud_cover]
})

prediction = rf_model.predict(new_data)

print("New Weather Conditions:")
print(new_data.to_string(index=False))
print("\nPredicted Solar Energy Generation:",
      round(prediction[0], 2))
```
